```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.axes_grid1 import host_subplot

data_name = 'precipitation.csv'

```

```python
disdro_df = pd.read_csv(data_name)
disdro_df
```

```python
values_plot = disdro_df.plot(x="Datetime", y="ADC Values")
values_plot.set_xlabel("Datetime")
values_plot.set_ylabel("Values")
values_plot.set_title("Values / Time")
plt.gcf().autofmt_xdate()
```

```python
drops_plot = disdro_df.plot(x="Datetime", y="Drops")
drops_plot.set_xlabel("Datetime")
drops_plot.set_ylabel("Drops")
drops_plot.set_title("Drops / Time")
plt.gcf().autofmt_xdate()
```
